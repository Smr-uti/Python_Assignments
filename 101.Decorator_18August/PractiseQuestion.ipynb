#### 1. Greeting decorator — Write a decorator greet that prints "Hello!" before running any function, and apply it to a function that prints your name.


```python
def greet(func):
    def inner():

        print("Hello")

        func()

    return inner

@greet
def my_name():
    print("My name is Smruti")

my_name()
```

#### 2. Add welcome message — Write a decorator that takes a function, runs it, and then prints "Thank you for using this program" after it finishes 


```python
def thank_you(func):
    def inner():
        func()
        print("Thank you for using this program")
    return inner

@thank_you
def welcome():
    print("Welcome in our home")

welcome()
```

#### 3. Simple math decorator — Write a decorator that takes a function returning a number and doubles the result before returning it (e.g., if the function returns 5, the decorator makes it return 10).

```python
print("Start Program")
def double(func):
    def inner():
        result = func()

        return result * 2
    return inner

@double
def add():
    return 5

print(add())
```
