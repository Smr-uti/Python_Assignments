#### Q8.  Write a class called Vehicle with:- __init__ method setting: self.color = "red" self.top_speed = 150 self.milage = 20 - A method display_car_details(self) that prints all three attributes Create an object obj1 = Vehicle() and call obj1.display_car_details() (Based on class1.py)


```python
class Vehicle:

      def __init__(self):
            self.color = "red"
            self.top_speed = 150
            self.mileage = 20

      def display_car_details(self):
            print("Color: ", self.color)
            print("Top_Speed: ", self.top_speed)
            print("Mileage: ",self.mileage)

obj1 = Vehicle()
obj1.display_car_details()
```

#### Q9. How many objects can you create from one class? Using the Vehicle class from Q8, create THREE separate objects: obj1, obj2, obj3. Call display_car_details() on each. Are the outputs the same? Why?


```python
class Vehicle:

      def __init__(self):
            self.color = "red"
            self.top_speed = 150
            self.mileage = 20

      def display_car_details(self):
            print("Color: ", self.color)
            print("Top_Speed: ", self.top_speed)
            print("Mileage: ",self.mileage)

obj1 = Vehicle()
obj2 = Vehicle()
obj3 = Vehicle()

obj1.display_car_details()
obj2.display_car_details()
obj3.display_car_details()
```
