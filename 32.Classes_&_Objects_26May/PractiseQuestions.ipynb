#### Q8.  Write a class called Vehicle with:- __init__ method setting: self.color = "red" self.top_speed = 150 self.milage = 20 - A method display_car_details(self) that prints all three attributes Create an object obj1 = Vehicle() and call obj1.display_car_details() (Based on class1.py)


```python
class Vehicle:

      def __init__(self):
            self.color = "red"
            self.top_speed = 150
            self.mileage = 20

      def display_car_details(self):
            print("Color: ", self.color)
            print("Top_Speed: ", self.top_speed)
            print("Mileage: ",self.mileage)

obj1 = Vehicle()
obj1.display_car_details()
```

#### Q12. Write a class called Student with: - Attributes: name, roll_no, marks - Method: show_info() that prints all three Create two objects with different attribute values and call show_info() on both.

```python
class Student:

      def __init__(self, name, roll_no, marks):
            self.name = name
            self.roll_no = roll_no
            self.marks = marks

      def show_info(self):
            print("Name: ", self.name)
            print("Roll No: ", self.roll_no)
            print("Marks: ", self.marks)

student1 = Student("Ramesh", 101, 90)
student2 = Student("Rajesh", 102, 85)

student1.show_info()
student2.show_info()

```

#### Q14. What is self in Python? Demonstrate by writing a class Box with: - __init__(self) setting width=10, height=5 - Method area(self) that returns width * height Create an object, call area(), and print the result.

```python
class Box:

    def __init__(self):
        self.width = 10
        self.height = 5

    def area(self):

        return self.width * self.height

obj =Box()

result = obj.area()

print("Area of the Box: ", result)
```

#### Q15. Write a class Calculator with: - __init__ setting self.result = 0 - Method add(self, a, b) that prints a + b - Method multiply(self, a, b) that prints a * b Create one object and call both methods.

```python
class Calculator:

    def __init__(self):
        self.result = 0

    def add(self, a, b):
        self.result = a + b
        print("Addition: ", self.result)

    def multiply(self, a, b):
        self.result = a * b
        print("Multiplication: ", self.result)

calc = Calculator()

calc.add(5, 10)
calc.multiply(5, 10)

```

#### Q17. Write a class Person with: - Attributes: name, age, city - Method greet(self) that prints: "Hello, my name is <name>, I am <age> years old and I live in <city>." Create two Person objects with different values and call greet() on each.

```python
class Person:

      def __init__(self, name, age, city):
            self.name = name
            self.age = age
            self.city = city

      def greet(self):
            print("Hello, my name is", self.name, "I am", self.age, "years old", "and I live in", self.city)

person1 = Person("Rahul", 30, "Mumbai")
person2 = Person("Mahesh", 25, "Pune")

person1.greet()
person2.greet()
```
