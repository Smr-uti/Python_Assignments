#### 1	Check iterable vs iterator — Take a list [1, 2, 3], print its type, then convert it to an iterator using iter() and print that type too.

```python
numbers = [1, 2, 3] # list is iterable, but list_iterator is an iterator.

print(type(numbers))

iterator = iter(numbers)

print(type(iterator))
```

#### 2. Manual iteration — Create an iterator from a list and call next() on it three times to print each element one by one, then show what happens when you call next() a fourth time.


```python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))

print(next(iterator)) #The 4th next() fails because the iterator has no more elements. StopIteration means "iteration is finished; no more elements."
```

#### 3	Memory comparison — Write two small programs: one using a list comprehension [x for x in range(1, 10001)] and another using range(1, 10001), and compare their memory size using sys.getsizeof().

```python
import sys

my_list = [x for x in range(1, 10001)]

print("List memory:", sys.getsizeof(my_list), "bytes")
```

```python
import sys

my_range = range(1, 10001)

print("Range memory:", sys.getsizeof(my_range), "bytes")
```
